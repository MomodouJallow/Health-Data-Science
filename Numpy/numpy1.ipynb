# Specifying the Data Type

```python
c = np.array([1, 2, 3, 4], dtype=float)
print(c)
```

```python
c.dtype
```

```python
z = np.array((1, 2, 4)).astype(float)
print(z)
```

# PLACE HOLDERS IN NUMPY ARRAYS

```python
# Create an array of zeros
x = np.zeros((4,5))
print(x)
```

```python
# Create an array of ones
y = np.ones((3,3))
print(y)
```

```python
# Create an array of specific value
z = np.full((5,4),5)
print(z)
```

```python
# Create an Identity Matrix
a = np.eye(4)
print(a)
```

```python
# Create random numbers between 0-1
b = np.random.random((3,4))
print(b)
```

```python
# Create Random Integer Values
c = np.random.randint(10, 100, (3,5)) # start, stop, (r, c)
print(c)
```

```python
# Create a numpy array of evenly spaced values
d = np.linspace(10, 30, 5)
print(d)
```
